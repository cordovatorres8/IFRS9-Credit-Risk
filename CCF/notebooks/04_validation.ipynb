## Protocolo cuando exista dataset adecuado
1. Validar fechas, definiciones y saldo/límite positivos. 2. Excluir denominadores nulos o bajo umbral predefinido y reportarlos. 3. Mantener LEQ sin winsorización en análisis descriptivo; hacer sensibilidad de colas documentada para el modelo. 4. Holdout temporal y agrupado por deudor. 5. Comparar media/mediana por segmento contra un modelo; MAE/RMSE, sesgo agregado, calibración por bandas y proporción de predicciones fuera del rango económico. 6. Reportar coverage y exclusiones.
